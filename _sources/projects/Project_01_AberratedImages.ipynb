## <span style="color:Orange">Overview</span>

Over the past 20–30 years the groundwork has been laid for precise experimental control of atomic gases at ultracold temperatures. Ultracold atom experiments exploit the quantum mechanics of their underlying atomic systems for a diverse set of applications, ranging from the simulation of computationally difficult physics problems [[1]](https://doi.org/10.1126/science.1209019) to searches for new physics beyond the Standard Model [[2]](https://doi.org/10.1103/PhysRevLett.119.153001). Most experiments in this field rely on directly imaging the ultracold gas, to extract the size and shape of the atomic number density distribution.

The goal of this project is to introduce you to some relevant image-processing techniques, and to get familiar with an image as a data element. As you will see, images are a kind of data with a very large number of features, almost all of which are highly correlated within a region of interest. Of particular interest is how a real imaging system distorts the information contained in an image, and how those effects can be unfolded from the data to recover information about the true density distribution.

Capturing all possible kinds of aberrations and noise in real images of ultracold atoms is outside the scope of the simulated data in this project. Instead, we limit ourselves to a few key effects:

* optical aberrations, in the form of defocus and primary spherical aberration;
* pixelization from the finite resolution of the detector; and
* in some datasets, a simple model of number noise.

All images simulate a thermal gas of $^{87}$Rb atoms imaged with $\lambda = 780$ nm light onto a camera with pixel size $\text{PS} = 13$ µm. The true parameters of each image are encoded in its file name, e.g. `HiNA.atom_num=100000.mu=(0.0...0.0...0.0).sigma=(1.0...1.0...20.0).Z04=0.2.seed=0.tiff`: the imaging system, the atom number $N$, the centre $\mu$ and widths $\sigma = (\sigma_x, \sigma_y, \sigma_z)$ of the Gaussian density in µm (with $z$ the imaging axis), the spherical-aberration coefficient `Z04`, and the random seed of the noise.

| Dataset | Imaging system | Images | Image size | Gas widths $\sigma$ (µm) | Atom numbers |
|---|---|---|---|---|---|
| `Number_estimate` | Low NA: NA = 0.2, magnification M = 4, no spherical aberration (`Z04` = 0) | 60 | 6 × 6 pixels | (10, 10, 10) | 9,000 / 10,000 / 11,000 and 99,000 / 100,000 / 101,000; 10 noise seeds each |
| `Heavy_tails` | High NA: NA = 0.4, M = 20, `Z04` = 0.2 | 18 | 30 × 30 pixels | $\sigma_{xy}$ ∈ {1, 2.5, 5, 10, 20}, with $\sigma_z = \sigma_{xy}$ or $\sigma_z = 20$ | 100,000; 2 noise seeds each |

**Noise model.** Every image is normalized so that its pixels sum to the atom number $N$. Then noise is added to each pixel with a standard deviation equal to the square root of the pixel value (see `ImSim` in `ImageSimulation.ipynb`). This is a toy model for fluctuations in the measured density. It can produce small negative pixel values, which were deliberately left in.

**Field of view.** The simulation grid extends only ±10 µm in each direction, transverse and axial. Gases that are large compared with this, such as the $\sigma = 10$–20 µm gases, are therefore truncated. You will look at the consequences in Question 05.

```python
%matplotlib inline
import glob
import os
import re
import tarfile
import urllib.request

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import skimage.io

DATA_URL = "https://courses.physics.illinois.edu/phys498mlp/sp2025/data/projects/AberratedImages/data.tgz"

# Download and extract the data once
if not os.path.isdir("Number_estimate"):
    urllib.request.urlretrieve(DATA_URL, "data.tgz")
    with tarfile.open("data.tgz") as tar:
        try:
            tar.extractall(".", filter="data")   # safe extraction (Python >= 3.12, and recent 3.8-3.11)
        except TypeError:
            tar.extractall(".")

PATTERN = re.compile(r"(?P<system>\w+)\.atom_num=(?P<N>\d+)\."
                     r"mu=\((?P<mu_x>[-\d.]+?)\.\.\.(?P<mu_y>[-\d.]+?)\.\.\.(?P<mu_z>[-\d.]+?)\)\."
                     r"sigma=\((?P<sigma_x>[\d.]+?)\.\.\.(?P<sigma_y>[\d.]+?)\.\.\.(?P<sigma_z>[\d.]+?)\)\."
                     r"Z04=(?P<Z04>[\d.]+)\.seed=(?P<seed>\d+)\.tiff")

def load_dataset(dataset_name):
    """Load all images in a dataset folder into a DataFrame: one row per image,
    with the true parameters parsed from the file name and the image itself in column 'image'."""
    rows = []
    for path in sorted(glob.glob(os.path.join(dataset_name, "*.tiff"))):
        m = PATTERN.fullmatch(os.path.basename(path))
        row = {k: float(v) for k, v in m.groupdict().items() if k != "system"}
        row["system"] = m["system"]
        row["image"] = skimage.io.imread(path)
        rows.append(row)
    df = pd.DataFrame(rows)
    return df.astype({"N": int, "seed": int})

number_df = load_dataset("Number_estimate")
tails_df = load_dataset("Heavy_tails")
print(len(number_df), "Number_estimate images of shape", number_df.image[0].shape)
print(len(tails_df), "Heavy_tails images of shape", tails_df.image[0].shape)
tails_df.drop(columns="image").head()
```

### <span style="color:LightGreen">Question 03</span>

Real imaging systems have a finite resolution, set by the diffraction limit. Consider a point source emitting a spherical electromagnetic wave. A real lens propagates the field from the object plane to the image plane, producing a pattern called the imaging response function, or [point spread function (PSF)](https://en.wikipedia.org/wiki/Point_spread_function), of the system. Any object of finite size can be thought of as a collection of point sources, so its image is the sum of the PSFs of all its points, i.e. a convolution.

For a circular lens, the PSF at best focus, for a point source on the optical axis, is the [Airy disk](https://en.wikipedia.org/wiki/Airy_disk), $I(\rho) = I_0 \left[2J_1(\rho)/\rho\right]^2$. Here $\rho = 2\pi\,\text{NA}\, r/\lambda$ is a dimensionless radial coordinate, and its first zero is at $\rho \approx 3.83$.

1. Calculate the fraction of the total irradiance (power) contained in the central peak, inside the first zero.
2. Approximate the Airy disk by a Gaussian, $I_0 \exp(-2\rho^2/w^2)$, with the same peak irradiance *and* the same total power. How is the first zero of the Airy disk related to the $1/e^2$ radius $w$ of the Gaussian? Compare with a least-squares Gaussian fit to the central peak.
3. Convert the radius of the first zero to µm for the two imaging systems used in this project (NA = 0.4 and NA = 0.2, $\lambda = 780$ nm).

In addition to the links above, a useful resource is Goodman's *Introduction to Fourier Optics* [[4]](#references).
